- ※ 강의 교안의 코드 블록을 원문 그대로 옮긴 것(핵심부 발췌 포함)

```python
thdot = np.linalg.pinv(J) @ xdot          # (가) 순정 pinv
# vs
def dls(J, xdot, lam=0.05):
    JT = J.T                               # (나) damped least squares
    return JT @ np.linalg.solve(J @ JT + lam**2 * np.eye(J.shape[0]), xdot)
```
